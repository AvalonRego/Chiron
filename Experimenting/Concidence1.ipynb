```python
from ananke.visualisation.detector import get_detector_scatter3ds
from ananke.models.collection import Collection
from ananke.configurations.collection import HDF5StorageConfiguration
import matplotlib.pyplot as plt
import numpy as np
```

```python
path='data/test/MIX.h5'
```

```python
config=HDF5StorageConfiguration(data_path=path,read_only=False)
```

```python
collection=Collection(config)
```

```python
collection.open()
detector=collection.storage.get_detector()
hits=collection.storage.get_hits()
records=collection.storage.get_records()
sources=collection.storage.get_sources()
collection.close()
```

```python
x=[]
for j,i in enumerate(records.df['record_id']):
    print('record :',i)
    plt.figure().set_figwidth(15)
    #plt.subplot(1, 2, 1)
    hit=hits.get_by_record_ids(i).df['time']
    #source=sources.get_by_record_ids(i).df['time']
    #a=plt.hist(source,bins=1)
    #a=1
    plt.title('sources vs time')
    #plt.yscale('log')
    #plt.subplot(1, 2, 2)
    plt.title('hits vs time')
    b=plt.hist(hit,bins=1)
    #plt.yscale('log')
    x.append([b])
plt.show()
```

```python
x=[]
for j,i in enumerate(records.df['record_id']):
    print('record :',i)
    plt.figure().set_figwidth(15)
    plt.subplot(1, 2, 1)
    hit=hits.get_by_record_ids(i).df['time']
    source=sources.get_by_record_ids(i).df['time']
    a=plt.hist(source,bins=1)
    #a=1
    plt.title('sources vs time')
    #plt.yscale('log')
    plt.subplot(1, 2, 2)
    plt.title('hits vs time')
    b=plt.hist(hit,bins=1)
    #plt.yscale('log')
    x.append([a,b])
plt.show()
```
