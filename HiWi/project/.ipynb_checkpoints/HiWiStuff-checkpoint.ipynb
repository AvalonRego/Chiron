```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# assume matrix is called data (shape: (472300, 2))
# col 0 = IDs, col 1 = frequencies
freq = data[:, 1]

# --- Basic Stats ---
count = freq.size
mean = np.mean(freq)
median = np.median(freq)
std = np.std(freq)
_min = np.min(freq)
_max = np.max(freq)
q25, q75 = np.percentile(freq, [25, 75])
mode = stats.mode(freq, keepdims=True)[0][0]
skewness = stats.skew(freq)
kurt = stats.kurtosis(freq)

print(f"Count: {count}")
print(f"Mean: {mean:.4f}")
print(f"Median: {median:.4f}")
print(f"Std: {std:.4f}")
print(f"Min: {_min}, Max: {_max}")
print(f"25%: {q25}, 75%: {q75}")
print(f"Mode: {mode}")
print(f"Skewness: {skewness:.4f}, Kurtosis: {kurt:.4f}")

# --- Plots ---
plt.figure(figsize=(14, 10))

# Histogram
plt.subplot(2, 2, 1)
sns.histplot(freq, bins=50, kde=False)
plt.title("Histogram of Frequencies")

# Boxplot
plt.subplot(2, 2, 2)
sns.boxplot(x=freq)
plt.title("Boxplot of Frequencies")

# CDF
plt.subplot(2, 2, 3)
sorted_freq = np.sort(freq)
cdf = np.arange(1, count + 1) / count
plt.plot(sorted_freq, cdf)
plt.title("CDF of Frequencies")

# Log-Log Histogram
plt.subplot(2, 2, 4)
plt.hist(np.log10(freq), bins=50, log=True)

plt.title("Log-Log Histogram")

plt.tight_layout()
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# assume matrix is called data (shape: (472300, 2))
# col 0 = IDs, col 1 = frequencies
freq = data[:, 1]

# --- Basic Stats ---
count = freq.size
mean = np.mean(freq)
median = np.median(freq)
std = np.std(freq)
_min = np.min(freq)
_max = np.max(freq)
q25, q75 = np.percentile(freq, [25, 75])
mode = stats.mode(freq, keepdims=True)[0][0]
skewness = stats.skew(freq)
kurt = stats.kurtosis(freq)

print(f"Count: {count}")
print(f"Mean: {mean:.4f}")
print(f"Median: {median:.4f}")
print(f"Std: {std:.4f}")
print(f"Min: {_min}, Max: {_max}")
print(f"25%: {q25}, 75%: {q75}")
print(f"Mode: {mode}")
print(f"Skewness: {skewness:.4f}, Kurtosis: {kurt:.4f}")

# --- Plots ---
plt.figure(figsize=(14, 10))

# Histogram
plt.subplot(2, 2, 1)
sns.histplot(freq, bins=50, kde=False)
plt.title("Histogram of Frequencies")

# Boxplot
plt.subplot(2, 2, 2)
sns.boxplot(x=freq)
plt.title("Boxplot of Frequencies")

# CDF
plt.subplot(2, 2, 3)
sorted_freq = np.sort(freq)
cdf = np.arange(1, count + 1) / count
plt.plot(sorted_freq, cdf)
plt.title("CDF of Frequencies")

# Log-Log Histogram
plt.subplot(2, 2, 4)
plt.hist(freq, bins=50, log=True)
plt.yscale("log")
plt.xscale("log")
plt.title("Log-Log Histogram")

plt.tight_layout()
plt.show()

```
